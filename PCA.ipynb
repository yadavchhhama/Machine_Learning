```python
import matplotlib.pyplot as plt
import seaborn as sns 
import numpy as np
import pandas as pd 
%matplotlib inline
```

```python
#load data 
from sklearn.datasets import load_breast_cancer
```

```python
cancer_dataset = load_breast_cancer()
```

```python
cancer_dataset.keys()
```

```python
type(cancer_dataset)
```

```python
print(cancer_dataset.DESCR)
```

```python
df = pd.DataFrame(cancer_dataset['data'], columns=cancer_dataset.feature_names)
```

```python
df.head()
```

```python
df.info()
```

```python
#we will try to extract two features (2D) that captures the maximum amount of information out of it for that we are using standardization 
from sklearn.preprocessing import StandardScaler 
scaler = StandardScaler()
```

```python
scaler.fit(df)
```

```python
scaled_data = scaler.transform(df)
```

```python
scaled_data
```

```python
##apply pca algorithm
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components = 3)
```

```python
data_pca = pca.fit_transform(scaled_data)
```

```python
data_pca
```

```python
pca.explained_variance_
```

```python
plt.figure(figsize=(12,10))
plt.scatter(data_pca[:,0],data_pca[:,1],data_pca[:,2], c=cancer_dataset['target'],cmap='plasma')
plt.xlabel('First principal component')
plt.ylabel('Second principal component')
plt.zlabel('Third principal component')
```
