```python
import seaborn as sns
import kagglehub
path = kagglehub.dataset_download("heptapod/titanic")
```

```python
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
#checking null value
df.isnull().sum()
```

```python
df.shape
```

```python
#when you delete columns wise
df.dropna(axis=1)
```

```python
#imputaton missing value
#1. mean value imputation :- it work well when we have normally distributed data
sns.displot(df['age'])
```

```python
sns.histplot(df['age'],kde=True)
```

```python
df['age_mean'] = df['age'].fillna(df['age'].mean())
```

```python
df[['age_mean','age']]
```

```python
sns.histplot(df['age_mean'])
```

```python
#2.Median value imputation :- if we have outliers in the dataset
df['age_median'] = df['age'].fillna(df['age'].median())
```

```python
df[['age_median','age_mean','age']]
```
